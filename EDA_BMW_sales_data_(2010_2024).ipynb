## Data Visualization - Distributions

### Subtask:
Generate histograms for numerical columns (e.g., 'Engine_Size_L', 'Mileage_KM', 'Price_USD', 'Sales_Volume') to visualize their distributions. Create bar plots for categorical columns (e.g., 'Model', 'Region', 'Color', 'Fuel_Type', 'Transmission', 'Sales_Classification') to show their frequencies.


**Reasoning**:
Generate histograms for the specified numerical columns to visualize their distributions, which is the first part of the subtask. I will use `plt.figure(figsize=...)` to control the plot size and `sns.histplot` for the histograms to ensure readability.



````python
numerical_cols = ['Engine_Size_L', 'Mileage_KM', 'Price_USD', 'Sales_Volume']

plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols, 1):
    plt.subplot(2, 2, i)
    sns.histplot(df[col], kde=True, bins=30)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
````

**Reasoning**:
Generate bar plots for the categorical columns 'Region', 'Color', 'Fuel_Type', 'Transmission', and 'Sales_Classification' to display the frequency of each category, as per the subtask instructions.



````python
categorical_cols = ['Region', 'Color', 'Fuel_Type', 'Transmission', 'Sales_Classification']

plt.figure(figsize=(20, 15))
for i, col in enumerate(categorical_cols, 1):
    plt.subplot(3, 2, i) # Adjust subplot grid for 5 plots
    sns.countplot(data=df, x=col, palette='viridis')
    plt.title(f'Frequency of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability
plt.tight_layout()
plt.show()
````

**Reasoning**:
The previous code generated a FutureWarning because `palette` was used without assigning a `hue` in `sns.countplot`. To address this warning and align with newer seaborn versions, I will explicitly set `hue=col` and `legend=False` within the `sns.countplot` function for the categorical plots.



````python
categorical_cols = ['Region', 'Color', 'Fuel_Type', 'Transmission', 'Sales_Classification']

plt.figure(figsize=(20, 15))
for i, col in enumerate(categorical_cols, 1):
    plt.subplot(3, 2, i) # Adjust subplot grid for 5 plots
    sns.countplot(data=df, x=col, hue=col, palette='viridis', legend=False)
    plt.title(f'Frequency of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability
plt.tight_layout()
plt.show()
````

**Reasoning**:
Now, I will generate a bar plot for the 'Model' column, specifically showing the frequency of the top 10 models to address the last instruction of the subtask. This will provide insights into the most popular BMW models.



````python
plt.figure(figsize=(12, 6))
top_10_models = df['Model'].value_counts().head(10)
sns.barplot(x=top_10_models.index, y=top_10_models.values, palette='viridis')
plt.title('Top 10 Most Frequent BMW Models')
plt.xlabel('Model')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
````

**Reasoning**:
The previous code generated a FutureWarning because `palette` was used without assigning a `hue` in `sns.barplot`. To address this warning and align with newer seaborn versions, I will explicitly set `hue=top_10_models.index` and `legend=False` within the `sns.barplot` function.



````python
plt.figure(figsize=(12, 6))
top_10_models = df['Model'].value_counts().head(10)
sns.barplot(x=top_10_models.index, y=top_10_models.values, hue=top_10_models.index, palette='viridis', legend=False)
plt.title('Top 10 Most Frequent BMW Models')
plt.xlabel('Model')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
````

## Sales Trends Analysis by Year

### Subtask:
Analyze the sales trends over time by visualizing 'Sales_Volume' against 'Year'.


**Reasoning**:
To analyze sales trends over time, I will first group the data by 'Year' and sum the 'Sales_Volume' for each year. This will provide the total sales volume for each year, which is necessary for visualizing the trend.



````python
sales_by_year = df.groupby('Year')['Sales_Volume'].sum().reset_index()
print("Sales Volume by Year:")
print(sales_by_year.head())
````

**Reasoning**:
Now that the sales volume by year has been aggregated, I will create a line plot to visualize the annual sales trend as per the subtask instruction.



````python
plt.figure(figsize=(12, 6))
sns.lineplot(data=sales_by_year, x='Year', y='Sales_Volume', marker='o')
plt.title('Annual Sales Volume Trend')
plt.xlabel('Year')
plt.ylabel('Total Sales Volume')
plt.grid(True)
plt.xticks(sales_by_year['Year'].unique(), rotation=45)
plt.tight_layout()
plt.show()
````

## Top Selling Models and Regions

### Subtask:
Identify and visualize the top-selling car models and regions based on 'Sales_Volume' using appropriate bar charts. Ensure plots include legends for clarity.


**Reasoning**:
To identify the top-selling car models, I will first group the DataFrame `df` by the 'Model' column and sum the 'Sales_Volume' for each model. Then, I will sort these results in descending order to easily pick the top performers.



````python
top_models_sales = df.groupby('Model')['Sales_Volume'].sum().sort_values(ascending=False)
print("Top Selling Models by Sales Volume:")
print(top_models_sales.head(10))
````

**Reasoning**:
Now that the top selling models by sales volume have been identified, I will create a bar chart to visualize the sales volume of these top models as specified in the subtask. I will visualize the top 10 models.



````python
plt.figure(figsize=(14, 7))
sns.barplot(x=top_models_sales.head(10).index, y=top_models_sales.head(10).values, hue=top_models_sales.head(10).index, palette='viridis', legend=False)
plt.title('Top 10 Selling BMW Models by Sales Volume')
plt.xlabel('Model')
plt.ylabel('Total Sales Volume')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
````

**Reasoning**:
To identify the top-selling regions, I will group the DataFrame `df` by the 'Region' column and sum the 'Sales_Volume' for each region, then sort the results in descending order.



````python
top_regions_sales = df.groupby('Region')['Sales_Volume'].sum().sort_values(ascending=False)
print("Top Selling Regions by Sales Volume:")
print(top_regions_sales.head())
````

**Reasoning**:
Now that the top-selling regions by sales volume have been identified, I will create a bar chart to visualize the sales volume for these regions as specified in the subtask instructions.



````python
plt.figure(figsize=(12, 6))
sns.barplot(x=top_regions_sales.index, y=top_regions_sales.values, hue=top_regions_sales.index, palette='viridis', legend=False)
plt.title('Sales Volume by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales Volume')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
````

## Outlier Detection

### Subtask:
Identify potential outliers in numerical columns like 'Price_USD', 'Mileage_KM', 'Engine_Size_L', and 'Sales_Volume' using box plots. Discuss their potential impact on the analysis.


**Reasoning**:
I need to generate box plots for the numerical columns 'Price_USD', 'Mileage_KM', 'Engine_Size_L', and 'Sales_Volume' to identify potential outliers, as per the subtask instruction. I will loop through these columns and create a separate box plot for each.
